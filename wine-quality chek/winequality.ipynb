```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
import joblib

# Load dataset
dataset = pd.read_csv('winequality-red.csv')

# Create label
dataset["quality_label"] = np.where(dataset["quality"] >= 6.5, "good", "rest")

# Features and target
data = dataset.iloc[:, 0:11].values  # 11 columns
target = dataset["quality_label"].values

# Split data
train_data, test_data, train_target, test_target = train_test_split(data, target, test_size=0.2, random_state=42)

# Random Forest model
model = RandomForestClassifier(n_estimators=100, criterion='gini', random_state=42)
model.fit(train_data, train_target)

# Predict
predicted_target = model.predict(test_data)

# Accuracy
print("Accuracy:", accuracy_score(test_target, predicted_target))

# Save model
joblib.dump(model, 'trained_wine_rf.sav')
```
