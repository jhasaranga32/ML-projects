```python
import pandas as pd
dataset=pd.read_csv('irish.csv',header=None,names=['sepal.len','sepal.width','petal.len','petal.width','variety'])
```

```python
print(dataset['variety'].unique())
dataset["variety"]=dataset["variety"].map({"Iris-setosa":0,"Iris-versicolor":1,"Iris-virginica":2})
dataset.head()
```

```python
data=dataset.values[:,0:4]
print(data)
```

```python
target=dataset.values[:,4]
```

```python
print(target,target.shape)
```

```python
from sklearn.model_selection import train_test_split
```

```python
train_data,test_data,train_target,test_target=train_test_split(data,target,test_size=0.2)
```

```python
print(test_data,test_data.shape)
```

```python
from sklearn.neighbors import KNeighborsClassifier
model=KNeighborsClassifier()
#model=KNeighborsClassifier(n_neighbors=3) /*using for change K value*/
```

```python
model.fit(train_data,train_target)
KNeighborsClassifier()
```

```python
predicted_target=model.predict(test_data)
print('predicted_target:',predicted_target)
```

```python
print(test_target)
```

```python
from sklearn.metrics import accuracy_score
acc=accuracy_score(test_target,predicted_target)
print('acc:',acc)
```

```python
predicted_train_target=model.predict(train_data)
print('train acc:', acc)
```
