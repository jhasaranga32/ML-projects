```python
import numpy as np
import joblib

from tensorflow.keras.models import load_model

# load model
model = load_model("concrete-quality.keras")

# load scalers
scaler_data = joblib.load("scaler_data.pkl")
scaler_target = joblib.load("scaler_target.pkl")
```

```python
# Example input
input_data = np.array([[ 
198.6,132.4,0,192,0,978.4,825.5,3,

]])
```

```python
input_scaled = scaler_data.transform(input_data)
prediction_scaled = model.predict(input_scaled)
prediction_real = scaler_target.inverse_transform(prediction_scaled)

print("Predicted Concrete Strength:", prediction_real[0][0])

pred = scaler_target.inverse_transform(
    model.predict(
        scaler_data.transform(input_data)
    )
)

print(f"Predicted Strength: {pred[0][0]:.2f} MPa")

```
